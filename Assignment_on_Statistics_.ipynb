1. Calculate the mean, median, and mode for the math, reading, and writing scores.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/IIT Palakkad Advanced AI/StudentsPerformance.csv')

# Calculate statistics
for subject in ['math score', 'reading score', 'writing score']:
    print(f"\n{subject.upper()}")
    print("Mean:", df[subject].mean())
    print("Median:", df[subject].median())
    print("Mode:", df[subject].mode().tolist())
```

2. Determine the range, variance, and standard deviation for each score category.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/IIT Palakkad Advanced AI/StudentsPerformance.csv')

# Calculate range, variance, and standard deviation
for subject in ['math score', 'reading score', 'writing score']:
    score_range = df[subject].max() - df[subject].min()
    variance = df[subject].var()      # sample variance
    std_dev = df[subject].std()       # sample standard deviation

    print(f"\n{subject.upper()}")
    print("Range:", score_range)
    print("Variance:", round(variance, 2))
    print("Standard Deviation:", round(std_dev, 2))
```

3. Identify the mean, median, and mode of math scores for students with different
parental education levels.

```python
import pandas as pd

# Load dataset
df = pd.read_csv('/content/drive/MyDrive/IIT Palakkad Advanced AI/StudentsPerformance.csv')

# Group by parental education and calculate statistics
grouped = df.groupby('parental level of education')['math score'].agg(['mean', 'median'])
modes = df.groupby('parental level of education')['math score'] \
          .agg(lambda x: x.mode().tolist())

# Combine results
summary = grouped.copy()
summary['mode'] = modes

print(summary)
```

10. Z-Test: Assume the national average math score is 50. Use a one-sample z-test to test
if this sample's average math score is significantly different from the national
average.

```python
import pandas as pd
from math import sqrt, erf

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/IIT Palakkad Advanced AI/StudentsPerformance.csv')

# Sample statistics
n = len(df['math score'])
mean_math = df['math score'].mean()
std_math = df['math score'].std(ddof=1)

# Hypothesized mean
mu0 = 50

# Z-statistic
z = (mean_math - mu0) / (std_math / sqrt(n))

# Standard normal CDF
def norm_cdf(x):
    return (1.0 + erf(x / sqrt(2))) / 2.0

# Two-tailed p-value
p_value = 2 * (1 - norm_cdf(abs(z)))

mean_math, std_math, n, z, p_value
print(f"\n")
print("mean_math:", mean_math)
print("std_math:", std_math)
print("Sample statistics:", n)
print("Z-statistic:", z)
print("Two-tailed p-value:", p_value)
```

11. T-Test: Compare the reading scores between students with different lunch types
(free/reduced vs. standard) to test if there is a significant difference.

```python
import pandas as pd
from scipy import stats

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/IIT Palakkad Advanced AI/StudentsPerformance.csv')

# Separate reading scores by lunch type
reading_free = df[df['lunch'] == 'free/reduced']['reading score']
reading_standard = df[df['lunch'] == 'standard']['reading score']

# Calculate group means
mean_free = reading_free.mean()
mean_standard = reading_standard.mean()

# Perform Welch’s t-test (unequal variances)
t_stat, p_value = stats.ttest_ind(reading_free, reading_standard, equal_var=False)

mean_free, mean_standard, t_stat, p_value
print(f"\n")
print("mean_free:", mean_free)
print("mean_standard:", mean_standard)
print("t_stat:", t_stat)
print("Z-statistic:", z)
print("p_value:", p_value)
```

12. F-Test: Check if the variances of math scores differ significantly between students
with and without test preparation.

```python
import pandas as pd
from scipy import stats

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/IIT Palakkad Advanced AI/StudentsPerformance.csv')

# Separate math scores by test preparation status
math_none = df[df['test preparation course'] == 'none']['math score']
math_completed = df[df['test preparation course'] == 'completed']['math score']

# Sample variances
var_none = math_none.var(ddof=1)
var_completed = math_completed.var(ddof=1)

# Sample sizes
n_none = len(math_none)
n_completed = len(math_completed)

# F-statistic (larger variance in numerator)
if var_none >= var_completed:
    F_stat = var_none / var_completed
    df1, df2 = n_none - 1, n_completed - 1
else:
    F_stat = var_completed / var_none
    df1, df2 = n_completed - 1, n_none - 1

# Two-tailed p-value
p_value = 2 * (1 - stats.f.cdf(F_stat, df1, df2))

var_none, var_completed, F_stat, p_value
```
