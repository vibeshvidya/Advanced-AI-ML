Weather Data Analyzer

```python
#Store Weather Analysis
#Store data in list
temps=[32,30,36,34,38]
humidity=[70,80,65,60,90]
```

```python
#Function to check the weather condition

def check_weather(temp,hum):
  if temp>30:
    return("Sunny")
  elif hum>80:
    return("Rainy")
  elif temp<25:
    return("Cool")
  else:
    return("Cloudy")
```

```python
#Function to calculate average

def average(data):
  return sum(data)/len(data)
```

```python
#Print results

for i in range(len(temps)):
  condition = check_weather(temps[i],humidity[i])
  print(f"Day{i+1}: Temps={temps[i]}, Humidity = {humidity[i]}% -> {condition}")
```

```python
print("\nAverage Temperature:", average(temps))
print("Average Humidity:", average(humidity))
```
