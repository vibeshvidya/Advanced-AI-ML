# Libraries

```python
import numpy as np
import pandas as pd

# Encoding.
from sklearn.preprocessing import LabelEncoder
# splitting data into train and test data
from sklearn.model_selection import train_test_split

# classification algorithms
from sklearn.linear_model import LogisticRegression # logistic regression
from sklearn.neighbors import KNeighborsClassifier  # KNN
from sklearn.svm import SVC                         # SVM
from sklearn.tree import DecisionTreeClassifier     # Decision Tree

# evaluation libraries
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
```

# Reading the dataset

```python
filepath = '/content/drive/MyDrive/ICT_Online/AI ML/IIT Palakkad Advanced AI/Data/iris.csv'
df_iris = pd.read_csv(filepath)
df_iris.head(2)
```

```python
df_iris.info()
```

# Pre processing.

```python
# before encoding,
df_iris['species'].unique()
```

```python
# Iris dataset is already pre processed.
# Except target column Encoding.

# since it is target column, we use label encoding
label_encode = LabelEncoder()
df_iris['species'] = label_encode.fit_transform(
                                  df_iris['species']
                                                  )
df_iris.head(2)
```

```python
# after encoding,
df_iris['species'].unique()
```

```python
# verify all the data in the dataFrame are numerical
# this needs to be done before ML model building.
```

# ML model building.

## Data splitting

```python
# % of data split among train and test data
# train 70 - 80
# test 30 - 20

# Split dataset in to target and features

y = df_iris['species']                  # target
X = df_iris.drop(['species'], axis =1)  # features
```

```python
y.head(2)
```

```python
X.head(2)
```

```python
# split into train and test datasets.
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state = 12)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
y_train.head()
```

```python
y_test.head()
```

## ML models

### Logistic Regression

```python
# initialize your model
log_model = LogisticRegression()
# train your model
log_model.fit(X_train, y_train)
# make predictions using the model
y_pred_log = log_model.predict(X_test)
# evaluate your model by comparing it's prediction with test values
acc_log = accuracy_score(y_pred_log, y_test)
prec_log = precision_score(y_pred_log, y_test,
                           average = 'weighted')
recall_log = recall_score(y_pred_log, y_test,
                          average = 'weighted')
f1_log = f1_score(y_pred_log, y_test,
                  average = 'weighted')
conf_matrix_log = confusion_matrix(y_pred_log, y_test)

print('Accuracy of the model : ', acc_log)
print('Precision of the model : ', prec_log)
print('Recall of the model : ', recall_log)
print('f1 score of the model : ', f1_log)
print('confusion matrix of the model : ', conf_matrix_log)
```

### k-NN
k Nearest Neighbours

```python
# assume k = 3
# initialize your model
knn_model = KNeighborsClassifier(n_neighbors=3)
# train your model
knn_model.fit(X_train, y_train)
# make predictions using the model
y_pred_knn = knn_model.predict(X_test)
# evaluate your model by comparing it's prediction with test values
acc_knn = accuracy_score(y_pred_knn, y_test)

print('Accuracy of the model : ', acc_knn)
```
