 Bank Account Manager

```python
#Simulate a bank account with deposit, withdrawal, and balance
# enquiry operations

# Account Information (Variables)

account_holder='Vibesh Vidya'
account_number='016450010620'
balance=1000.00
```

```python
#Function: Deposit Money
def deposit_money(amount):
  global balance
  if amount>0:
    balance+=amount
    print(f"Amount deposited: {amount}")
    print(f"Balance Amount: {balance}")
  else:
    print("Invalid Amount")

print("\n--- Deposit Operation ---")
deposit_money(100)
```

```python
#Function: Withdraw Money

def Withdraw(amount):
  global balance
  if amount <= balance: # Corrected condition for valid withdrawal
    balance -= amount
    print(f"Amount withdrawal: {amount}")
    print(f"Balance Amount: {balance}")
  elif amount <= 0:
    print("Invalid Withdrawal Amount. Please enter a positive amount.")
  else: # amount > balance
    print("Insufficient funds for withdrawal.")

print("\n--- Withdrawal Operation ---") # Changed print statement for clarity
Withdraw(100)
```

```python
# Function: Check Balance

def check_balance():
  print("Account Holder name ", account_holder)
  print("Account Number ",account_number)
  print("Balance amount", balance)

check_balance()
```
