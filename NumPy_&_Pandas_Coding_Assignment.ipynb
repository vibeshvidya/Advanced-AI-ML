**Employee Salary Adjustment**

```python
#1. Employee Salary Adjustment

import pandas as np
salaries = np.array([45000, 52000, 61000, 48000, 75000])
print(salaries)
print('-----------------------------------------')

#Increase all salaries by 10%

salary_increment = salaries * 1.1
print("Increase in Salary",salary_increment)
print('-----------------------------------------')

#Find the Maximum salary.

print("Maximum Salary:", max(salaries))
print('-----------------------------------------')

#Find the minimum salary.
print('Minimum Salary:',min(salaries))
print('-----------------------------------------')

#Print salaries greater than 60000.
higher_salary = salaries[salaries>60000]
print('Salary > 60000', higher_salary)
print('-----------------------------------------')
```

**Sales Data Analysis**

```python
from numpy import average
import pandas as np
sales = np.array([120,150,180,200,170,160,210,220,190,175])

#1. Compute total sales.

print("Total Sales : ",sales.sum())
print("---------------------------------------------------------")

#2. Compute mean and standard deviation.

print("Mean : ",sales.mean())
print("---------------------------------------------------------")
print("Standard Deviation: ",sales.std())
print("-----------------------------------")
#3. Identify days where sales exceeded the average.

exceed_sales= sales>sales.mean()
print("Sales exceed the average",exceed_sales)
print("---------------------------------------------------------")
#4. Normalize the sales using Min-Max scaling.

print("Minimum Sales: ",sales.min())
print("---------------------------------------------------------")
print("Minimum Sales: ",sales.max())
print("---------------------------------------------------------")
min_normal = sales-sales.min()
max_normal = sales-sales.max()
normalized_sales = (min_normal)/ (max_normal)
print("Normalized Sales : ", normalized_sales)
```

**Matrix Operations for Business Data**

```python
import numpy as np

sales_matrix = np.array([
[120,140,130,150],
[80,90,100,95],
[200,210,190,205]
])

#1. Find total sales per product.

total_sales = np.sum(sales_matrix,axis=1)
print("Total Sales per product: ",total_sales)

#2. Find total sales per region.

total_sales = np.sum(sales_matrix,axis=0)
print("Total Sales per region: ",total_sales)

#3. Identify the best performing product.

best_product = np.argmax(sales_matrix,axis=1)
print("Best product: Product",best_product+1)

#4. Identify the region with highest sales.
highest_sales = np.argmax(sales_matrix,axis=0)
print("Region with Highest Sales: Sales ",highest_sales+1)
```

**Missing Data Handling**

```python
from numpy._core.fromnumeric import mean
import numpy as np
import pandas as pd

data = {
'Name':['John','Anna','Peter','Linda','David'],
'Age':[28, np.nan, 35, 32, np.nan],
'Salary':[50000,54000,np.nan,58000,60000]
}

# 1. Detect missing values.

df = pd.DataFrame(data)
print("Missing Values: ",df.isnull)

# 2. Replace missing Age with mean age.

mean_age = df['Age'].mean()

#df['Age'].fillna(mean_age, inplace=True)
df['Age'] = df['Age'].fillna(mean_age)
df['Age']


# 3. Replace missing Salary with median salary.

mean_salary = df['Salary'].mean()

df['Salary'] = df['Salary'].fillna(mean_salary)
df['Salary']


# 4. Remove rows with any remaining missing values.

df.dropna(inplace=True)

print('Cleaned Data: \n',df)
```
